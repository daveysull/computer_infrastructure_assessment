#### Meanings of $ in Bash Scripts

1) data_dir="data/weather"
mkdir -p "$data_dir"

In this scenario $ is used to access the value stored in a variable. data/weather is replaced by $data_dir. This increases efficiency as it allows changes to be made in one place and it will update in the script. mkdir -p is a function we used frequently to create a directory. Using double quotes around $data_dir ensures it works properly even if the folder name has spaces.

2) timestamp=$(date +"%Y%m%d_%H%M%S")

In the code above \$ is used instead of backticks which we used frequently throughout the semester. Backticks and \$ functionally do the same thing however I decided to stick with \$ for readability.  


3) output_file="\$data_dir/\${timestamp}.json"

\$ is used here to combine variables into a single string. \$data_dir and \$timestamp are replaced by their values, creating a full file path. This shows how $ helps make file names and paths dynamic and reusable.

4) echo "Weather data saved to $output_file"

In this line, $output_file is replaced by the actual file name and path when the script runs. This makes the message automatically show the correct file name without you needing to type it manually.

```python
#!/bin/bash

# Define the directory to save the weather data
data_dir="data/weather"

# Ensure the directory exists
mkdir -p "$data_dir"

# Get the current timestamp in the format YYYYmmdd_HHMMSS
timestamp=$(date +"%Y%m%d_%H%M%S")

# Define the URL for the weather data
weather_url="https://prodapi.metweb.ie/observations/athenry/today"

# Define the output file name with the timestamp
output_file="$data_dir/${timestamp}.json"

# Download the weather data and save it to the timestamped file
wget -O "$output_file" "$weather_url"

# Print a success message
echo "Weather data saved to $output_file"
```

#### Getting Started with GitHub Actions

After completing weather.sh I was able to begin the final project. For ease of use the project was done through codespace in GitHub. In our week 9 lecture we were shown GitHub Co Pilot which automates the bash script that we want. Creating new bash scripts isn't most people do frequently so we were encouraged to automate the script if possible. Unfortunately co-pilot wasn't working for me so I used ChatGPT to create the below script. While automating the script is extremely helpful it's equally important to understand how it works. For this reason I've provided a breakdown of what the script is doing below.

1) name: This is used to give our workflow a name which is this scenario is "New Weather Data".

2) on, schedule: Is used to run the workflow on a schedule, in this example it runs every day at 11AM

3) workflow_dispatch : Allows the workflow to be triggered manually from GitHub Actions

4) jobs: Is used to define the task/job to be performed in the workflow and "runs-on:" specifies the environment where the job will run.

5) uses: actions/checkout@v3 Gets the repository code so that the workflow can access necessary files

6) git config: Is used to set the name and email for commits made by this worflow, identifying them as made by GitHub Actions bot.

7) chmod +x weather.sh : Ensures that the script is executable. Once this is done ./weather.sh runs the script previously created.

8) The final part of the code is adding all changes, committing and pushing them.

Being able to automate this script made life a lot easier but still needed some adjustments to ensure it ran correctly.

#### References

https://docs.github.com/en/actions/writing-workflows/workflow-syntax-for-github-actions


```python
name: New Weather Data

on:
  schedule:
    - cron: '0 11 * * *' 
  workflow_dispatch:

jobs:
  update-weather-data:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout Repository
        uses: actions/checkout@v3
      - name: Configure Git User
        run: |
          git config --global user.name "github-actions[bot]"
          git config --global user.email "github-actions[bot]@users.noreply.github.com"
      - name: Run weather.sh
        run: |
          chmod +x weather.sh
          ./weather.sh
      - name: Commit and Push Changes
        run: |
          git add .
          git commit -m "Update weather data [GitHub Actions]"
          git push
```
