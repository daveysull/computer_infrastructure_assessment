#### Task 4: Create Timestamped Files

In this task I used two functions I knew already which were Date and Touch. What was new in this task was combining them and the use of backticks.
Whatever is inbetween the backticks is executed first before the main command so in this example we are getting the date in the format we want first and then it runs the touch command.
Syntax: touch `date "+%Y%m%d_%H%M%S"`.txt
